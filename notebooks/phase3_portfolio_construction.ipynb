# Phase 3 — Portfolio Construction
Runs the portfolio modules and displays results. All logic lives in `src/portfolio/`, this notebook only calls it.

```python
import sys, os
sys.path.append(os.path.abspath('..'))
import pandas as pd

from src.data_collection import config
from src.data_collection.saver import save_processed
from src.portfolio.weights import equal_weights, user_defined_weights, market_cap_weights
from src.portfolio.construction import build_portfolio
```

```python
# Load Phase 2 output (cleaned, winsorized prices)
prices = pd.read_csv(config.DATA_PROCESSED_DIR / 'prices_clean_phase2.csv', index_col='date', parse_dates=True)
prices.tail()
```

```python
# Option 1: Equal-weighted portfolio
eq_weights = equal_weights(list(prices.columns))
eq_result = build_portfolio(prices, eq_weights, initial_value=1_000_000.0)
eq_result.value.tail()
```

```python
# Option 2: User-defined weights (example split — edit to your own view)
custom_weights = user_defined_weights({
    'AAPL': 0.15, 'MSFT': 0.15, 'NVDA': 0.15, 'JPM': 0.10, 'XOM': 0.05,
    'AMZN': 0.10, 'META': 0.10, 'GOOGL': 0.10, 'TSLA': 0.05, 'UNH': 0.05,
})
custom_result = build_portfolio(prices, custom_weights, initial_value=1_000_000.0)
custom_result.value.tail()
```

```python
# Option 3: Market-cap weighted portfolio (live lookup — needs internet)
mcap_weights = market_cap_weights(list(prices.columns))
mcap_result = build_portfolio(prices, mcap_weights, initial_value=1_000_000.0)
mcap_result.value.tail()
```

```python
# Monthly-rebalanced version of the equal-weighted portfolio, for comparison
eq_rebal_result = build_portfolio(prices, eq_weights, initial_value=1_000_000.0, rebalance_freq='ME')
print(f"Rebalances occurred on: {eq_rebal_result.rebalance_dates}")
eq_rebal_result.value.tail()
```

```python
# Save Phase 3 outputs (using the equal-weighted, non-rebalanced portfolio as the primary output)
save_processed(eq_result.value.to_frame(), 'portfolio_value_equal_weighted.csv', config.DATA_PROCESSED_DIR)
save_processed(eq_result.pnl.to_frame(), 'portfolio_pnl_equal_weighted.csv', config.DATA_PROCESSED_DIR)
save_processed(eq_result.returns.to_frame(), 'portfolio_returns_phase3.csv', config.DATA_PROCESSED_DIR)
print('Phase 3 outputs saved.')
```
